## Задача 1. Один крок для однієї машини (Datsun PL510)

```python
x = 0.97
y = 8.7
w = 5
lr = 0.1

pred = w * x
error = (pred - y) ** 2
print("прогноз:", pred)
print("помилка:", error)

grad = 2 * (pred - y) * x
print("крутизна:", grad)

w_new = w - lr * grad
pred_new = w_new * x
error_new = (pred_new - y) ** 2
print("нове w:", w_new)
print("новий прогноз:", pred_new)
print("нова помилка:", error_new)
print("Стало краще?", error_new < error)
```

## Задача 2. Чи згодні машини?

```python
cars = {"Fiat 128": (0.85, 8.1), "Pontiac Catalina": (2.01, 16.8)}

def grad(w, x, y):
    return 2 * (w * x - y) * x

for w in (5, 9):
    print("w =", w)
    for name, (x, y) in cars.items():
        print(" ", name, grad(w, x, y))

for name, (x, y) in cars.items():
    print(name, "ідеальне w = y / x =", y / x)
```

При w = 5 обидві крутизни від'ємні, тому обидві машини радять збільшувати w (напрям однаковий).

При w = 9 крутизна Fiat від'ємна (збільшувати w), а Pontiac додатна (зменшувати w): напрями різні.

Причина: знак крутизни залежить від знака (прогноз − y), тобто від того, чи прогноз менший, чи більший за правильну відповідь. Ідеальне w для Fiat ≈ 9.53, для Pontiac ≈ 8.36. При w = 5 воно нижче обох ідеальних значень, тому обидві машини тягнуть w вгору. Значення w = 9 лежить між 8.36 і 9.53: для Fiat воно ще замале, а для Pontiac вже завелике.

## Задача 3. Розмір кроку

```python
def run(lr, steps=5, w=12):
    print("lr =", lr)
    for i in range(1, steps + 1):
        w = w - lr * 2 * (w - 8)
        print("  крок", i, "w =", w, "помилка =", (w - 8) ** 2)

for lr in (0.5, 0.9, 1.0, 1.1):
    run(lr)
```

Крок оновлення: w − 8 множиться на (1 − 2·lr) на кожному кроці.

- 0 < lr < 1: |1 − 2·lr| < 1, помилка зменшується (збіжність).
- lr = 1: множник дорівнює −1, w стрибає між 12 і 4, помилка завжди 16. Модель не збігається і не розбігається, а коливається.
- lr > 1: |1 − 2·lr| > 1, помилка зростає (модель «вилітає»).

Найбільше lr, за якого модель ще не вилітає, дорівнює 1.
